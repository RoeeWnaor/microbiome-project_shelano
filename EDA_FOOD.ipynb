### About the Data

This comprehensive dataset maps the microbial landscape of various foods, aggregated from 85 distinct scientific experiments. In total, the collection encompasses 4,344 unique food samples and profiles the abundance of approximately 14,000 different bacterial taxa. 

The data is organized into two primary, interconnected tables that separate the contextual information from the biological measurements:

* **The Metadata Table:** This table provides the descriptive, geographical, and technical context for every sample. Key columns include hierarchical food classifications (such as `macrocategory`, `category`, `type`, `subtype`, and `commercial_name`), the sample's geographical origin (`country`), and whether the food is fermented. It also contains essential technical metrics from the sequencing process, such as read lengths and base counts.
* **The Taxonomic Profile Table:** This table holds the actual microbiome composition data. It details the specific bacterial taxa—represented by complete taxonomic lineages from kingdom down to the species or strain level—and records their precise abundance or presence within the samples.

**Data Integration:** These two tables are intrinsically linked using a unique composite identifier. By combining the `dataset_name` with the specific `sample_id` from the metadata, we create a primary key that matches the sample columns/rows in the taxonomic profile. This seamless relationship allows us to correlate the physical and categorical attributes of any given food item directly with its complex microbial ecosystem.

### Let's check for null values existance:

````python
def plot_missing_values(df, title, ax):
    
    # Count null values for each column and keep only columns with missing values
    missing_data = df.isnull().sum()
    missing_data = missing_data[missing_data > 0].sort_values(ascending=False)
    
    if missing_data.empty:
        # If there are no missing values, display a message
        ax.text(0.5, 0.5, 'No Null Values Found! \n(Dataset is clean)', 
                ha='center', va='center', fontsize=14, color='green', fontweight='bold')
        ax.set_title(title, fontsize=14)
        ax.axis('off') # Hide the axes because there is no data to display
    else:
        # If there are missing values, create a bar plot
        # Show the top 20 columns with the most missing values to avoid overcrowding
        top_missing = missing_data.head(20) 
        sns.barplot(x=top_missing.values, y=top_missing.index, ax=ax, palette='magma')
        ax.set_title(title, fontsize=14)
        ax.set_xlabel('Number of Missing Values')
        ax.set_ylabel('Columns')

# Set up the visualization area with two plots side by side
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

print("Analyzing null values existence...")

# Apply the function to both master tables
plot_missing_values(df_master_metadata, "Missing Data in Metadata Table", axes[0])
plot_missing_values(df_master_tax_profile, "Missing Data in Tax Profile Table", axes[1])

# Adjust spacing automatically and display the plots
plt.tight_layout()
plt.show()
````

The missing values analysis shows that the unified taxonomic profile table contains no null values, which confirms that the taxonomic abundance matrix is complete after replacing missing entries with zero during the merging step. In contrast, the metadata table contains several columns with missing values, mainly commercial_name, collection_date, and experiment_accession, followed by smaller amounts of missingness in fields such as subtype, DNA_extraction_kit, and accession-related columns. ### Food Item Duplication and Consistency Analysis

When evaluating the food attributes in our metadata (specifically within columns such as `subtype`, `category`, or `commercial_name`), it is crucial to distinguish between two distinct types of "duplicates":

* **Valid Duplicates (Multiple Samples per Food Item):**
  This metric tracks how many times a specific food item (e.g., a particular cheese type in the `subtype` column or a specific brand in the `commercial_name` column) was actually sampled. Having multiple biological or technical samples of the same food is excellent for research validity and statistical power. However, it is essential to map out this distribution to identify any class imbalances in the dataset.

* **"Dirty" Duplicates (Data Entry Inconsistencies & Typos):**
  This occurs when the exact same food item is recorded under slightly different text formats across different rows. For example, entries might appear inconsistently as `"Cheddar"`, `"cheddar"`, `"Cheddar "` (with a trailing space), or `"Cheddar Cheese"`. Resolving these inconsistencies is a critical preprocessing step; if left uncleaned, a machine learning model will incorrectly interpret them as entirely distinct food features, which can severely degrade its predictive accuracy and pattern recognition.

````python

# Make sure the metadata table is available
df_meta = df_master_metadata.copy()

# Select the column that represents the specific food item.
# This can be changed to 'commercial_name' or 'category' depending on the desired resolution
food_column = 'subtype' 

print(f"--- Duplicate analysis for column: '{food_column}' ---")

# Remove missing values
foods_series = df_meta[food_column].dropna().astype(str)

# 1. Which food items appear most often? Multiple samples
food_counts = foods_series.value_counts()
repeated_foods = food_counts[food_counts > 1]

print(f"\nThere are {len(repeated_foods)} food types that appear more than once in the dataset.")
if not repeated_foods.empty:
    print("Top 10 most frequent food types:")
    display(pd.DataFrame({'Count': repeated_foods.head(10)}))

# Create a quick plot of the most repeated food items
plt.figure(figsize=(10, 5))
sns.barplot(x=repeated_foods.head(15).values, y=repeated_foods.head(15).index, palette='mako')
plt.title(f'Top 15 Most Repeated Foods ({food_column})')
plt.xlabel('Number of Samples')
plt.show()

# ---------------------------------------------------------
# 2. Detect dirty duplicates: typos, case sensitivity, and extra spaces
# ---------------------------------------------------------
print("\n--- Searching for dirty text duplicates, meaning the same food written in different ways ---")

# Create a cleaned version of the text: lowercase letters and no extra spaces at the beginning or end
clean_foods = foods_series.str.lower().str.strip()

# Check which original values were mapped to the same cleaned value
# This can reveal cases such as "Cheddar" and "cheddar"
mapping_df = pd.DataFrame({
    'Original': foods_series,
    'Cleaned': clean_foods
})

# Group by the cleaned value and check how many different original values it has
dirty_duplicates = mapping_df.groupby('Cleaned')['Original'].unique()

# Keep only values that have more than one original spelling version
inconsistent_spellings = dirty_duplicates[dirty_duplicates.apply(len) > 1]

if not inconsistent_spellings.empty:
    print(f"Found {len(inconsistent_spellings)} food items with inconsistent spelling.")
    print("Examples of inconsistencies that should be reviewed and possibly standardized:")
    for clean_name, original_versions in inconsistent_spellings.head(10).items():
        print(f"  * The food item '{clean_name}' appears as: {list(original_versions)}")
else:
    print("No capitalization or extra-space spelling inconsistencies were found. The records are consistent.")
````

### Taxonomic Composition and Microbial Profiling

To make sense of the highly dimensional microbial data, we aggregate the complex taxonomic strings into higher-level biological classifications. This allows us to understand the general "bacterial families" dominating our food samples.

* **Macro-Level Distribution (Phylum):** The pie chart illustrates the average relative abundance of the major bacterial phyla (e.g., Firmicutes, Proteobacteria, Bacteroidota) across all samples. This represents the high-level microbial landscape of the dataset.
* **Micro-Level Prevalence (Genus):** The bar chart displays the top 15 most prevalent bacterial genera. Rather than looking at abundance, this metric counts in how many distinct food samples a specific genus (e.g., *Lactobacillus*, *Lactococcus*) appears, highlighting the most universal microbes in our food database.

````python

# 1. Detect orientation: get the taxa strings and the data
if df_tax.shape[0] > df_tax.shape[1] or any('Bacteria' in str(idx) for idx in df_tax.index[:5]):
    # Taxa are rows
    taxa_strings = pd.Series(df_tax.index)
    abundance_data = df_tax
else:
    # Taxa are columns
    taxa_strings = pd.Series(df_tax.columns)
    abundance_data = df_tax.T

print("Parsing taxonomic hierarchy...")

# 2. Extract Phylum (p__) and Genus (g__) using Regex
phylum_series = taxa_strings.str.extract(r'p__([^|]+)')[0].fillna('Unknown')
genus_series = taxa_strings.str.extract(r'g__([^|]+)')[0].fillna('Unknown')


# --- ERROR FIX: Force data to be strictly numeric before calculating ---
# This converts any accidental strings into NaN, and then replaces them with 0
abundance_numeric = abundance_data.apply(pd.to_numeric, errors='coerce').fillna(0)


# --- Plot 1: Phylum Distribution (Pie Chart based on Mean Abundance) ---

# Calculate the mean abundance for each specific taxon across all samples using the numeric data
mean_abundance_per_taxon = abundance_numeric.mean(axis=1).values

# Create a dataframe to group by Phylum
phylum_df = pd.DataFrame({
    'Phylum': phylum_series,
    'Mean_Abundance': mean_abundance_per_taxon
})

# Sum abundance by Phylum
phylum_abundance = phylum_df.groupby('Phylum')['Mean_Abundance'].sum().sort_values(ascending=False)

# Group small phyla into "Others" to keep the pie chart clean
threshold = phylum_abundance.sum() * 0.02 # 2% threshold
main_phyla = phylum_abundance[phylum_abundance >= threshold]
other_phyla = pd.Series({'Others ( < 2%)': phylum_abundance[phylum_abundance < threshold].sum()})
pie_data = pd.concat([main_phyla, other_phyla])

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Draw Pie Chart
axes[0].pie(
    pie_data.values, 
    labels=pie_data.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=sns.color_palette("pastel"),
    wedgeprops={'edgecolor': 'black', 'linewidth': 1}
)
axes[0].set_title('Average Microbial Abundance by Phylum', fontsize=15, fontweight='bold', pad=20)


# --- Plot 2: Top Genera Prevalence (Bar Chart based on presence > 0) ---

# Calculate in how many samples each taxon appears (using the numeric data)
prevalence_per_taxon = (abundance_numeric > 0).sum(axis=1).values

# Create a dataframe to group by Genus
genus_df = pd.DataFrame({
    'Genus': genus_series,
    'Prevalence': prevalence_per_taxon
})

# Sum prevalence by Genus
genus_prevalence = genus_df.groupby('Genus')['Prevalence'].sum().sort_values(ascending=False)

# Remove 'Unknown' if it's there, and take top 15
if 'Unknown' in genus_prevalence:
    genus_prevalence = genus_prevalence.drop('Unknown')
top_genera = genus_prevalence.head(15)

# Draw Bar Chart
sns.barplot(x=top_genera.values, y=top_genera.index, ax=axes[1], palette='flare')
axes[1].set_title('Top 15 Most Prevalent Bacterial Genera\n(Total appearances across all samples)', fontsize=15, fontweight='bold', pad=20)
axes[1].set_xlabel('Total Detections', fontsize=12)
axes[1].set_ylabel('')

# Add labels to the bars
for i, v in enumerate(top_genera.values):
    axes[1].text(v + (max(top_genera.values) * 0.01), i, str(int(v)), color='black', va='center')

sns.despine(ax=axes[1])
plt.tight_layout()
plt.show()
````

### Microbial Diversity and Dominance (Alpha Diversity)

To understand the internal microbial ecosystem of our food samples, we evaluate two critical diversity metrics across different food categories. These metrics help us differentiate between complex, diverse microbiomes and highly specialized (or targeted) environments.

* **Microbial Richness:** The total number of unique bacterial taxa detected (abundance > 0) within a single food sample. High richness indicates a complex and highly diverse microbial community.
* **Microbial Dominance:** The maximum abundance level of the single most prevalent bacterial taxon in a sample. High dominance suggests that the sample's microbiome is heavily skewed and controlled by one specific bacterium (a common characteristic in strictly controlled fermented foods).

````python
# Ensure the dataframes are available
df_meta = df_metadata if 'df_metadata' in locals() else df_master_metadata
df_tax = df_tax_profile if 'df_tax_profile' in locals() else df_master_tax_profile

print("Calculating alpha diversity metrics...")

# --- THE FIX: Ensure we have a matching ID column in metadata ---
# Recreate the 'combined_id' (e.g., LeechJ_2020__ERS3967688) so the tables can talk to each other
if 'combined_id' not in df_meta.columns:
    if 'dataset_name' in df_meta.columns and 'sample_id' in df_meta.columns:
        df_meta['combined_id'] = df_meta['dataset_name'] + '__' + df_meta['sample_id'].astype(str)

# 1. Detect orientation: Ensure rows are samples and columns are taxa
if df_tax.shape[0] > df_tax.shape[1] or any('Bacteria' in str(idx) for idx in df_tax.index[:5]):
    abundance_data = df_tax.T
else:
    abundance_data = df_tax

# Force data to be strictly numeric
abundance_numeric = abundance_data.apply(pd.to_numeric, errors='coerce').fillna(0)

# 2. Calculate Metrics per Sample
richness = (abundance_numeric > 0).sum(axis=1)
dominance = abundance_numeric.max(axis=1)

metrics_df = pd.DataFrame({
    'Richness': richness,
    'Dominance': dominance
})

# 3. Merge with Metadata to group by food category
food_group_column = 'category' # You can change this to 'subtype' or 'type'

# Set the correct index for merging
if 'combined_id' in df_meta.columns:
    df_meta_temp = df_meta.set_index('combined_id')
elif 'sample_id' in df_meta.columns:
    df_meta_temp = df_meta.set_index('sample_id')
else:
    df_meta_temp = df_meta.copy()

# Join the metrics with the selected food grouping column
plot_data = metrics_df.join(df_meta_temp[[food_group_column]], how='inner')

# Drop missing categories
plot_data = plot_data.dropna(subset=[food_group_column])

# --- Debug Print to ensure the merge worked ---
print(f"Successfully matched {len(plot_data)} samples between Metadata and Tax Profile!")
if len(plot_data) == 0:
    print("Warning: 0 samples matched. Please check if your Sample IDs match exactly.")

# 4. Generate Visualizations
if len(plot_data) > 0:
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # Plot 1: Richness Plot
    sns.boxplot(data=plot_data, x=food_group_column, y='Richness', ax=axes[0], palette='viridis')
    sns.stripplot(data=plot_data, x=food_group_column, y='Richness', ax=axes[0], color='black', alpha=0.3, size=3)
    axes[0].set_title('Microbial Richness per Food Category\n(Number of Unique Taxa)', fontsize=14, fontweight='bold', pad=15)
    axes[0].tick_params(axis='x', rotation=45)
    axes[0].set_xlabel('')
    axes[0].set_ylabel('Number of Unique Bacteria')

    # Plot 2: Dominance Plot
    sns.boxplot(data=plot_data, x=food_group_column, y='Dominance', ax=axes[1], palette='plasma')
    sns.stripplot(data=plot_data, x=food_group_column, y='Dominance', ax=axes[1], color='black', alpha=0.3, size=3)
    axes[1].set_title('Microbial Dominance per Food Category\n(Abundance of the Top Taxon)', fontsize=14, fontweight='bold', pad=15)
    axes[1].tick_params(axis='x', rotation=45)
    axes[1].set_xlabel('')
    axes[1].set_ylabel('Max Abundance Level')

    sns.despine()
    plt.tight_layout()
    plt.show()

    # 5. Print a numerical summary table
    print("-" * 50)
    print(f"Average Metrics per Food Group ('{food_group_column}'):")
    print("-" * 50)
    summary_stats = plot_data.groupby(food_group_column).mean().round(2)
    display(summary_stats)
````

### Extreme Values and Outlier Analysis (Alpha Diversity)

Beyond average metrics, identifying extreme values in microbial distribution provides critical insights into atypical samples within our dataset. 

* **High Extremes (Richness):** Samples with exceptionally high numbers of unique taxa may indicate highly complex natural fermentations, diverse environmental exposure, or potential contamination.
* **High Extremes (Dominance):** Samples where a single taxon approaches 100% abundance indicate a hyper-specialized environment, often seen in highly controlled monoculture fermentations or extreme microbial selection.
* **Diversity Landscape:** By plotting Richness against Dominance, we can visualize the typical ecological trade-off (highly diverse samples are rarely dominated by one microbe) and easily spot structural outliers that defy this trend.

````python
print("Analyzing extreme values for alpha diversity...\n")

# --- 1. Quick Data Prep (Ensuring standalone execution) ---
if 'combined_id' not in df_meta.columns:
    if 'dataset_name' in df_meta.columns and 'sample_id' in df_meta.columns:
        df_meta['combined_id'] = df_meta['dataset_name'] + '__' + df_meta['sample_id'].astype(str)

if df_tax.shape[0] > df_tax.shape[1] or any('Bacteria' in str(idx) for idx in df_tax.index[:5]):
    abundance_numeric = df_tax.T.apply(pd.to_numeric, errors='coerce').fillna(0)
else:
    abundance_numeric = df_tax.apply(pd.to_numeric, errors='coerce').fillna(0)

# Calculate metrics
richness = (abundance_numeric > 0).sum(axis=1)
dominance = abundance_numeric.max(axis=1)
metrics_df = pd.DataFrame({'Richness': richness, 'Dominance': dominance})

# Set index for merging
if 'combined_id' in df_meta.columns:
    df_meta_temp = df_meta.set_index('combined_id')
elif 'sample_id' in df_meta.columns:
    df_meta_temp = df_meta.set_index('sample_id')
else:
    df_meta_temp = df_meta.copy()

# Join with metadata columns to see WHAT food is an extreme outlier
cols_to_keep = [col for col in ['category', 'subtype', 'commercial_name'] if col in df_meta_temp.columns]
extreme_data = metrics_df.join(df_meta_temp[cols_to_keep], how='inner')

# --- 2. Extract and Display Extreme Values (Tables) ---
print("=" * 60)
print("EXTREME VALUES ANALYSIS (TOP & BOTTOM 5)")
print("=" * 60)

print("\nTop 5 HIGHEST Richness (Most diverse / potentially contaminated):")
display(extreme_data.nlargest(5, 'Richness'))

print("\nTop 5 LOWEST Richness (Least diverse / highly sterile):")
display(extreme_data.nsmallest(5, 'Richness'))

print("\nTop 5 HIGHEST Dominance (Most heavily dominated by a single taxon):")
display(extreme_data.nlargest(5, 'Dominance'))


# --- 3. Visualization: Scatter Plot of the Landscape ---
plt.figure(figsize=(12, 7))

# Use category for colors if available
hue_col = 'category' if 'category' in cols_to_keep else None

sns.scatterplot(
    data=extreme_data, 
    x='Richness', 
    y='Dominance', 
    hue=hue_col, 
    palette='tab10', 
    alpha=0.7, 
    edgecolor='black',
    s=70 # Dot size
)

plt.title('Microbial Diversity Landscape: Richness vs. Dominance\n(Outliers are isolated points far from the main clusters)', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Richness (Number of Unique Taxa)', fontsize=12)
plt.ylabel('Dominance (Max Abundance of Top Taxon)', fontsize=12)

# Move the legend outside the plot so it doesn't cover data points
if hue_col:
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title=hue_col.capitalize())

sns.despine()
plt.tight_layout()
plt.show()
````

## Step 1 — Remove Duplicated Microbiome Profile Columns

**What this step does:** Identifies sample columns in the taxonomic profile table that have 100% identical bacterial abundance vectors, and removes all but the first occurrence of each duplicated group.

**Why it is needed:** Duplicate samples inflate any statistical metric — diversity scores, abundance averages, and distance matrices — because the same biological signal is counted multiple times. Any downstream clustering or ordination would be unfairly pulled towards the duplicated samples.

**What problem it solves:** The EDA in the preceding section already detected 26 samples whose microbiome profiles are completely identical. This step removes those duplicates before any further analysis.

**What output to expect:** The number of columns in the taxonomic profile table decreases by the number of removed duplicates. A bar chart and a table confirm exactly which samples were removed.

**How this affects later stages:** All downstream analyses (matching, diversity, food-group comparisons) will work on a dataset free of replicated signal, making statistical results more reliable.

````python
print('=' * 60)
print('STEP 1: Remove Duplicated Microbiome Profile Columns')
print('=' * 60)

# Start from the raw table; samples = columns, taxa = index
tax_profile_clean = tax_profile_raw.copy()
n_original = tax_profile_clean.shape[1]

# Transpose so samples become rows — pandas duplicated() works on rows
tax_T    = tax_profile_clean.T
dup_mask = tax_T.duplicated(keep='first')
removed_samples = tax_T[dup_mask].index.tolist()
n_removed  = len(removed_samples)

# Drop the duplicate columns from the (non-transposed) table
tax_profile_clean = tax_profile_clean.loc[:, ~dup_mask]
n_remaining = tax_profile_clean.shape[1]

print(f'Original sample columns : {n_original}')
print(f'Duplicate columns removed: {n_removed}')
print(f'Remaining sample columns : {n_remaining}')

# ── Visualisation ────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, max(5, n_removed * 0.45 + 3)))

# Left: before/after bar chart
labels = ['Before\nRemoval', 'Duplicates\nRemoved', 'After\nRemoval']
values = [n_original, n_removed, n_remaining]
colors = ['#4C72B0', '#DD8452', '#55A868']
bars = axes[0].bar(labels, values, color=colors, edgecolor='black', width=0.5)
for bar, val in zip(bars, values):
    axes[0].text(bar.get_x() + bar.get_width() / 2,
                 bar.get_height() + n_original * 0.01,
                 str(val), ha='center', va='bottom',
                 fontweight='bold', fontsize=12)
axes[0].set_title('Sample Columns: Before vs. After\nDuplicate Removal',
                   fontsize=13, fontweight='bold')
axes[0].set_ylabel('Number of Sample Columns')
axes[0].set_ylim(0, n_original * 1.15)
sns.despine(ax=axes[0])

# Right: table of removed sample IDs
axes[1].axis('off')
if removed_samples:
    table_data = [[i + 1, s] for i, s in enumerate(removed_samples)]
    tbl = axes[1].table(
        cellText=table_data,
        colLabels=['#', 'Removed Duplicate Sample ID'],
        loc='center', cellLoc='left'
    )
    tbl.auto_set_font_size(False)
    tbl.set_fontsize(8)
    tbl.scale(1.2, 1.4)
    axes[1].set_title('Removed Duplicate Sample IDs',
                       fontsize=12, fontweight='bold', pad=20)
else:
    axes[1].text(0.5, 0.5, 'No duplicates found',
                  ha='center', va='center', fontsize=14)

plt.tight_layout()
plt.show()
````

## Step 2 — Match Metadata and Taxonomic Profile

**What this step does:** Constructs a composite sample key `combined_id = dataset_name + '__' + sample_id` in the metadata table and compares it against the column names of the taxonomic profile table to identify which samples appear in both, only in metadata, or only in the taxonomic profile.

**Why it is needed:** The two tables were generated independently and may not cover exactly the same set of samples. Proceeding with unmatched samples would introduce `NaN` values in any joint analysis and produce misleading results.

**What problem it solves:** Any sample that exists in one table but not the other cannot be used in downstream analyses that require both biological measurements and sample-level metadata. Removing unmatched samples ensures every sample has a complete record.

**What output to expect:** A bar chart compares the four key counts — total metadata samples, total taxonomic profile samples, matched, and unmatched. Both tables are then filtered to the matched set.

**How this affects later stages:** All downstream analyses (alpha diversity, food-group comparisons, PCA) will work on a perfectly aligned pair of tables where every sample has both a metadata row and a taxonomic profile column.

````python
print('=' * 60)
print('STEP 2: Match Metadata and Taxonomic Profile')
print('=' * 60)

metadata_clean = metadata_raw.copy()

# Build the composite key: must match the tax-profile column format
metadata_clean['combined_id'] = (
    metadata_clean['dataset_name'].astype(str) + '__' +
    metadata_clean['sample_id'].astype(str)
)

tax_ids  = set(tax_profile_clean.columns.tolist())
meta_ids = set(metadata_clean['combined_id'].tolist())

only_meta = meta_ids - tax_ids
only_tax  = tax_ids  - meta_ids
common    = meta_ids & tax_ids

print(f'Metadata samples              : {len(meta_ids)}')
print(f'Taxonomic profile samples     : {len(tax_ids)}')
print(f'Matched (both tables)         : {len(common)}')
print(f'Only in metadata (unmatched)  : {len(only_meta)}')
print(f'Only in tax profile (unmatched): {len(only_tax)}')

# Filter both tables to the matched set only
metadata_clean    = metadata_clean[
    metadata_clean['combined_id'].isin(common)
].reset_index(drop=True)

tax_profile_clean = tax_profile_clean.loc[
    :, tax_profile_clean.columns.isin(common)
]

print(f'\nAfter filtering to matched samples:')
print(f'  metadata_clean shape    : {metadata_clean.shape}')
print(f'  tax_profile_clean shape : {tax_profile_clean.shape}')

# ── Bar chart ────────────────────────────────────────────────────────────
labels_m = ['Metadata\nsamples', 'Tax-profile\nsamples',
             'Matched', 'Only in\nmetadata', 'Only in\ntax profile']
values_m = [len(meta_ids), len(tax_ids), len(common),
             len(only_meta), len(only_tax)]
colors_m = ['#4C72B0', '#4C72B0', '#55A868', '#DD8452', '#C44E52']

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(labels_m, values_m, color=colors_m, edgecolor='black', width=0.55)
for bar, val in zip(bars, values_m):
    ax.text(bar.get_x() + bar.get_width() / 2,
             bar.get_height() + max(values_m) * 0.01,
             str(val), ha='center', va='bottom', fontweight='bold', fontsize=11)
ax.set_title('Metadata vs. Taxonomic Profile — Sample Matching Summary',
              fontsize=13, fontweight='bold')
ax.set_ylabel('Number of Samples')
ax.set_ylim(0, max(values_m) * 1.15)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
````

If the number of matched samples is close to both the metadata total and the taxonomic profile total, the two tables are well-aligned and the small number of unmatched samples can safely be discarded. A large number of unmatched samples on either side would require investigation into how the files were generated before proceeding.

## Step 4 — Create the Unified `food_group` Column

**What this step does:** Combines `category`, `type`, and `subtype` into a single, interpretable column called `food_group` using a transparent rule-based string-matching function.

**Why it is needed:** The original dataset uses a four-level food hierarchy (`macrocategory > category > type > subtype`). For EDA and visualisation purposes, this is too granular — many unique combinations would produce charts that are impossible to read. Collapsing them into ~10 meaningful groups makes patterns visible.

**What problem it solves:** Without consolidation, any group-level plot would have dozens of tiny bars or wedges, making it impossible to identify which food types drive microbial diversity or abundance.

**What output to expect:** A new `food_group` column in `metadata_clean`. Three visualisations confirm the mapping: (1) a sample table showing original values alongside the assigned group, (2) a bar chart of samples per group, and (3) a table showing how many unique original combinations map to each group.

**How this affects later stages:** `food_group` will serve as the primary grouping variable in all subsequent diversity, abundance, and ordination analyses.

````python
print('=' * 60)
print('STEP 4: Create Unified food_group Column')
print('=' * 60)

def assign_food_group(row):
    s = ' '.join([
        str(row.get('category', '') or ''),
        str(row.get('type',     '') or ''),
        str(row.get('subtype',  '') or ''),
    ]).lower()

    if any(w in s for w in [
        'cheese','cheddar','mozzarella','brie','camembert',
        'ricotta','gouda','parmesan','feta','pecorino',
        'grana','raclette','manchego','gruyere'
    ]):
        return 'Cheese'
    if 'kefir' in s:
        return 'Kefir'
    if any(w in s for w in ['yogurt', 'yoghurt', 'skyr']):
        return 'Yogurt'
    if any(w in s for w in [
        'cocoa','cacao','chocolate','coffee',
        'fermented seed','fermented bean'
    ]):
        return 'Cocoa / fermented seeds'
    if any(w in s for w in [
        'sauerkraut','kimchi','pickle',
        'fermented vegetable','fermented veg','vegetable'
    ]):
        return 'Fermented vegetables'
    if any(w in s for w in [
        'beer','wine','beverage','juice','kombucha',
        'drink','cider','vinegar','spirit','fermented drink'
    ]):
        return 'Beverages'
    if any(w in s for w in [
        'fish','seafood','salmon','tuna','shrimp',
        'shellfish','prawn','squid','oyster','mussel','crab'
    ]):
        return 'Fish'
    if any(w in s for w in [
        'meat','beef','pork','poultry','chicken','turkey',
        'sausage','salami','ham','lamb','veal','prosciutto',
        'chorizo','bacon','mortadella','lard'
    ]):
        return 'Meat'
    if any(w in s for w in [
        'cow milk','bovine milk','raw milk','bulk milk'
    ]):
        return 'Cow milk'
    if any(w in s for w in [
        'milk','cream','butter','whey','dairy','colostrum','curd'
    ]):
        return 'Other dairy'
    return 'Other food'

metadata_clean['food_group'] = metadata_clean.apply(assign_food_group, axis=1)

fg_counts = metadata_clean['food_group'].value_counts()
print('Samples per food_group:')
print(fg_counts.to_string())

n_combos = metadata_clean[['category','type','subtype']].drop_duplicates().shape[0]
print(f'\nUnique category/type/subtype combinations: {n_combos}')

# ── Fig 1: mapping examples table ────────────────────────────────────────
example_rows = []
for grp in sorted(metadata_clean['food_group'].unique()):
    subs = (
        metadata_clean[metadata_clean['food_group'] == grp]
        [['category','type','subtype','food_group']]
        .drop_duplicates()
        .head(2)
    )
    example_rows.append(subs)
example_df = pd.concat(example_rows, ignore_index=True)

fig1, ax1 = plt.subplots(figsize=(14, len(example_df) * 0.45 + 1.5))
ax1.axis('off')
tbl_ex = ax1.table(
    cellText=example_df.values,
    colLabels=example_df.columns,
    loc='center', cellLoc='left'
)
tbl_ex.auto_set_font_size(False)
tbl_ex.set_fontsize(9)
tbl_ex.scale(1.2, 1.6)
ax1.set_title(
    'food_group Mapping — Examples (category | type | subtype → food_group)',
    fontsize=13, fontweight='bold', pad=20
)
plt.tight_layout()
plt.show()

# ── Fig 2: bar chart of samples per food_group ───────────────────────────
fig2, ax2 = plt.subplots(figsize=(10, 5))
sns.barplot(
    x=fg_counts.values, y=fg_counts.index, ax=ax2, palette='tab10'
)
for i, v in enumerate(fg_counts.values):
    ax2.text(v + max(fg_counts.values) * 0.01, i,
              str(v), va='center', fontsize=10)
ax2.set_title('Number of Samples per food_group',
               fontsize=13, fontweight='bold')
ax2.set_xlabel('Number of Samples')
ax2.set_ylabel('')
sns.despine(ax=ax2)
plt.tight_layout()
plt.show()

# ── Fig 3: unique original combinations per food_group ───────────────────
combo_counts = (
    metadata_clean
    .groupby('food_group')[['category','type','subtype']]
    .apply(lambda df: df.drop_duplicates().shape[0])
    .sort_values(ascending=False)
)
fig3, ax3 = plt.subplots(figsize=(10, 4))
sns.barplot(
    x=combo_counts.values, y=combo_counts.index, ax=ax3, palette='mako'
)
for i, v in enumerate(combo_counts.values):
    ax3.text(v + 0.1, i, str(v), va='center', fontsize=10)
ax3.set_title(
    'Unique Original category/type/subtype Combinations per food_group',
    fontsize=12, fontweight='bold'
)
ax3.set_xlabel('Number of Unique Combinations')
ax3.set_ylabel('')
sns.despine(ax=ax3)
plt.tight_layout()
plt.show()
````

## Step 5 — Drop Columns with Many Missing Values

**What this step does:** Removes eight columns that have high missing rates and were kept in Step 3 only until `food_group` was created. `subtype` is also evaluated: if more than 50% of its values are missing it is dropped, otherwise it is retained.

**Why it is needed:** Columns with very high proportions of missing values add noise to summary statistics and can distort imputation or model-based analyses if left in the dataset.

**What problem it solves:** Without this step the cleaned metadata would still contain columns where the majority of entries are `NaN`, which would appear prominently in any missing-value visualisation and mislead the analyst into thinking the data is incomplete in a general sense.

**What output to expect:** Two side-by-side bar charts show the missing-value profile *before* and *after* the drop. A comparison table lists every column that had missing values, its before/after status, and whether it was dropped or kept.

**How this affects later stages:** After this step the remaining missing values are limited to columns where the missingness is moderate and potentially informative (e.g., `country` or `subtype`), making the dataset much cleaner for EDA.

````python
print('=' * 60)
print('STEP 5: Drop High-Missing Columns (Before/After Comparison)')
print('=' * 60)

cols_to_drop = [
    'commercial_name', 'collection_date', 'experiment_accession',
    'DNA_extraction_kit', 'study_accession', 'run_accession',
    'project_accession', 'library_layout',
]

# ── Capture BEFORE state ─────────────────────────────────────────────────
miss_cnt_before = metadata_clean.isnull().sum()
miss_pct_before = (metadata_clean.isnull().mean() * 100).round(2)
miss_before_df  = pd.DataFrame({
    'Missing Count': miss_cnt_before,
    'Missing %'    : miss_pct_before,
}).query('`Missing Count` > 0').sort_values('Missing Count', ascending=False)

# ── Drop the columns ─────────────────────────────────────────────────────
present_drops = [c for c in cols_to_drop if c in metadata_clean.columns]
metadata_clean.drop(columns=present_drops, inplace=True)

subtype_status = 'not present'
subtype_dropped = False
if 'subtype' in metadata_clean.columns:
    sub_pct = metadata_clean['subtype'].isna().mean() * 100
    if sub_pct > 50:
        metadata_clean.drop(columns=['subtype'], inplace=True)
        subtype_status  = f'DROPPED ({sub_pct:.1f}% missing)'
        subtype_dropped = True
    else:
        subtype_status = f'KEPT ({sub_pct:.1f}% missing)'

print(f'Dropped {len(present_drops)} high-missing columns.')
print(f'subtype: {subtype_status}')
print(f'metadata_clean shape after dropping: {metadata_clean.shape}')

# ── Capture AFTER state ──────────────────────────────────────────────────
miss_cnt_after = metadata_clean.isnull().sum()
miss_pct_after = (metadata_clean.isnull().mean() * 100).round(2)
miss_after_df  = pd.DataFrame({
    'Missing Count': miss_cnt_after,
    'Missing %'    : miss_pct_after,
}).query('`Missing Count` > 0').sort_values('Missing Count', ascending=False)

# ── Side-by-side bar charts ──────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, max(5, len(miss_before_df) * 0.45 + 2)))

def _missing_bar(ax, df, title):
    if df.empty:
        ax.text(0.5, 0.5, 'No missing values!',
                 ha='center', va='center', fontsize=14,
                 color='green', fontweight='bold')
        ax.axis('off')
    else:
        sns.barplot(x=df['Missing %'].values, y=df.index, ax=ax, palette='magma')
        for i, v in enumerate(df['Missing %'].values):
            ax.text(v + 0.5, i, f'{v:.1f}%', va='center', fontsize=9)
        ax.set_xlabel('Missing (%)')
        sns.despine(ax=ax)
    ax.set_title(title, fontsize=12, fontweight='bold')

_missing_bar(axes[0], miss_before_df, 'Missing Values BEFORE\nDropping High-Missing Cols')
_missing_bar(axes[1], miss_after_df,  'Missing Values AFTER\nDropping High-Missing Cols')
plt.tight_layout()
plt.show()

# ── Comparison table ─────────────────────────────────────────────────────
all_dropped_set = set(present_drops)
if subtype_dropped:
    all_dropped_set.add('subtype')

cmp_rows = []
for col in miss_before_df.index:
    cmp_rows.append({
        'Column'            : col,
        'Missing Before'    : int(miss_before_df.loc[col, 'Missing Count']),
        'Missing % Before'  : f"{miss_before_df.loc[col, 'Missing %']:.1f}%",
        'Action'            : 'DROPPED' if col in all_dropped_set else 'KEPT',
    })
cmp_df = pd.DataFrame(cmp_rows)
print('\nBefore/After Comparison Table:')
display(cmp_df)
````

## Step 7 — Remove Non-Taxonomic Rows from the Taxonomic Profile

**What this step does:** Scans the row index of `tax_profile_clean` and removes any row whose index value is a metadata keyword (e.g., `category`, `country`) rather than a real taxonomic lineage string. A real taxon is identified by the presence of `k__` (kingdom prefix) in its name.

**Why it is needed:** When the taxonomic profile table was originally assembled, some metadata fields may have been accidentally appended as extra rows. These non-taxonomic rows would corrupt any calculation that treats every row as a bacterial taxon — for example, diversity metrics or abundance sums.

**What problem it solves:** Including metadata strings as if they were taxa would produce incorrect abundance values and confuse any analysis that relies on taxonomic hierarchies.

**What output to expect:** A before/after bar chart shows how many rows were removed. A small table lists the exact non-taxonomic row names that were found and deleted.

**How this affects later stages:** After this step, every row in `tax_profile_clean` represents a genuine microbial taxon, making the table safe to use for diversity analysis, ordination, and visualisation.

````python
print('=' * 60)
print('STEP 7: Remove Non-Taxonomic Rows from Tax Profile')
print('=' * 60)

n_rows_before = tax_profile_clean.shape[0]

non_tax_exact = {
    'macrocategory', 'category', 'type', 'subtype',
    'country', 'source', 'fermented', 'dataset_name',
    'fermented/non-fermented',
}
exact_mask  = tax_profile_clean.index.str.lower().isin(non_tax_exact)
no_kk_mask  = ~tax_profile_clean.index.str.contains('k__', na=False)
remove_mask = exact_mask | no_kk_mask

removed_rows = tax_profile_clean.index[remove_mask].tolist()
n_removed_r  = len(removed_rows)

tax_profile_clean = tax_profile_clean[~remove_mask]
n_taxa_remaining  = tax_profile_clean.shape[0]

print(f'Rows before cleaning    : {n_rows_before}')
print(f'Non-taxonomic rows found: {n_removed_r}')
print(f'Taxa remaining          : {n_taxa_remaining}')

# ── Visualisation ────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, max(5, n_removed_r * 0.45 + 3)))

bar_l = ['Before\nCleaning', 'Non-tax\nRemoved', 'Taxa\nRemaining']
bar_v = [n_rows_before, n_removed_r, n_taxa_remaining]
bar_c = ['#4C72B0', '#DD8452', '#55A868']
bars3 = axes[0].bar(bar_l, bar_v, color=bar_c, edgecolor='black', width=0.5)
for bar, val in zip(bars3, bar_v):
    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + n_rows_before * 0.01,
        str(val), ha='center', va='bottom', fontweight='bold', fontsize=12
    )
axes[0].set_title('Taxonomic Profile Rows:\nBefore vs. After Non-Tax Removal',
                   fontsize=13, fontweight='bold')
axes[0].set_ylabel('Number of Rows')
axes[0].set_ylim(0, n_rows_before * 1.15)
sns.despine(ax=axes[0])

axes[1].axis('off')
if removed_rows:
    tbl_r = axes[1].table(
        cellText=[[i + 1, r] for i, r in enumerate(removed_rows)],
        colLabels=['#', 'Removed Non-Taxonomic Row'],
        loc='center', cellLoc='left'
    )
    tbl_r.auto_set_font_size(False)
    tbl_r.set_fontsize(9)
    tbl_r.scale(1.2, 1.6)
    axes[1].set_title('Removed Non-Taxonomic Row Names',
                       fontsize=12, fontweight='bold', pad=20)
else:
    axes[1].text(0.5, 0.5, 'No non-taxonomic rows found',
                  ha='center', va='center', fontsize=12)

plt.tight_layout()
plt.show()
````

# Metadata Reduction, Table Merge, and Outlier Exploration

This section continues directly from the cleaned tables produced in the preprocessing section. The goals are:

1. Reduce the metadata to the minimum columns needed for EDA.
2. Transpose the taxonomic profile so that rows = samples.
3. Merge the two tables into a single analysis-ready DataFrame.
4. Identify informative taxa and explore potential outliers with scatter plots.

**Inputs required:** `metadata_clean`, `tax_profile_clean`  
**Key outputs:** `metadata_final`, `tax_profile_samples`, `merged_food_microbiome`

## Task 1 — Final Metadata Column Reduction → `metadata_final`

**What this step does:** Selects only the six most important descriptive columns from `metadata_clean` and stores them in a new DataFrame called `metadata_final`.

**Why it is needed:** The cleaned metadata still contains sequencing-quality metrics (`n_of_reads`, `mean_read_len`, etc.) and processing flags that are not relevant to the biological questions we want to explore. Keeping only the descriptive columns makes every subsequent join, plot, and table more readable.

**What output to expect:** `metadata_final` has 6 columns. A before/after bar chart confirms the reduction. A warning is printed if any required column is unexpectedly absent.

**How this helps the next stage:** The lean metadata table merges cleanly with the taxonomic profile without importing noise columns, and every downstream plot can use `food_group`, `country`, `type`, and `fermented/non-fermented` as grouping variables.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import itertools

print('=' * 60)
print('TASK 1: Final Metadata Column Reduction')
print('=' * 60)

REQUIRED_COLS = [
    'combined_id', 'sample_id', 'fermented/non-fermented',
    'type', 'country', 'food_group'
]

# Verify every required column is present
missing_req = [c for c in REQUIRED_COLS if c not in metadata_clean.columns]
if missing_req:
    print(f'WARNING: The following required columns are missing from metadata_clean:')
    for c in missing_req:
        print(f'  - {c}')
    available_cols = [c for c in REQUIRED_COLS if c in metadata_clean.columns]
    print('Proceeding with available columns only.')
else:
    available_cols = REQUIRED_COLS
    print('All required columns are present in metadata_clean.')

n_before = metadata_clean.shape[1]
metadata_final = metadata_clean[available_cols].copy()
n_after = metadata_final.shape[1]

print(f'\nColumns before: {n_before}')
print(f'Columns after:  {n_after}')
print(f'Columns dropped:{n_before - n_after}')
print(f'\nmetadata_final shape: {metadata_final.shape}')
display(metadata_final.head())

# Bar chart
fig, ax = plt.subplots(figsize=(7, 5))
bl = ['Before\nReduction', 'Columns\nDropped', 'After\nReduction']
bv = [n_before, n_before - n_after, n_after]
bc = ['#4C72B0', '#DD8452', '#55A868']
bars = ax.bar(bl, bv, color=bc, edgecolor='black', width=0.5)
for b, v in zip(bars, bv):
    ax.text(b.get_x() + b.get_width()/2, b.get_height() + 0.15,
             str(v), ha='center', va='bottom', fontweight='bold', fontsize=12)
ax.set_title('metadata_clean → metadata_final: Column Reduction',
              fontsize=13, fontweight='bold')
ax.set_ylabel('Number of Columns')
ax.set_ylim(0, n_before * 1.25)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
````

### Interpretation — Column Reduction

The six retained columns cover every dimension needed for exploratory analysis:

- `combined_id` — the primary join key linking metadata to the taxonomic profile.
- `sample_id` — human-readable sample label useful for annotating outliers.
- `food_group` — the consolidated food category; the main grouping variable.
- `type` — a finer food classification useful when `food_group` is too broad.
- `fermented/non-fermented` — a key biological variable expected to drive major microbiome differences.
- `country` — geographic origin, relevant for analysing regional variation.

All sequencing-quality metrics and technical identifiers have been removed because they describe the *measurement process*, not the *food sample* itself.

## Task 2 — Transpose Taxonomic Profile → `tax_profile_samples`

**What this step does:** The current `tax_profile_clean` table has taxa as rows and sample IDs as column names. This step transposes the table so that each *row* represents one food sample and each *column* represents one taxon. The result is stored as `tax_profile_samples`.

**Why it is needed:** Pandas `merge()` works on rows. To join the taxonomic abundance data with the metadata we need both tables to be sample-per-row. Transposing makes this possible and also puts the data in the standard 'observations × features' format expected by scikit-learn, SciPy, and most statistics libraries.

**What output to expect:** `tax_profile_samples` has the same number of rows as there are samples and the same number of columns as there are taxa, plus one `combined_id` column. Shapes before and after are printed.

**How this helps the next stage:** `tax_profile_samples` can be merged directly with `metadata_final` on `combined_id` to produce the single analysis-ready table used for all subsequent visualisations.

````python
print('=' * 60)
print('TASK 2: Transpose Taxonomic Profile')
print('=' * 60)

print(f'tax_profile_clean shape (taxa x samples): {tax_profile_clean.shape}')

# Transpose: taxa become columns, sample IDs become the index
tax_profile_samples = tax_profile_clean.T.copy()
tax_profile_samples = tax_profile_samples.reset_index()
# After reset_index the former index becomes a column; rename it combined_id
first_col = tax_profile_samples.columns[0]
tax_profile_samples.rename(columns={first_col: 'combined_id'}, inplace=True)

print(f'tax_profile_samples shape (samples x taxa): {tax_profile_samples.shape}')
print(f"First column (sample key): '{tax_profile_samples.columns[0]}'")
print(f'Sample combined_id values: {tax_profile_samples["combined_id"].head(3).tolist()}')

# Preview: first 5 rows, first 6 columns
display(tax_profile_samples.iloc[:5, :6])
````

## Task 3 — Merge Metadata and Taxonomic Profile → `merged_food_microbiome`

**What this step does:** Performs an inner join of `metadata_final` and `tax_profile_samples` on `combined_id` to create one unified DataFrame called `merged_food_microbiome`.

**Why it is needed:** Having metadata and abundance data in one table removes the need to do repeated index lookups during analysis. Every row in `merged_food_microbiome` is a complete food-sample record: descriptive metadata columns first, followed by the full set of taxonomic abundance values.

**What output to expect:** `merged_food_microbiome` has the same number of rows as the number of matched samples and the number of columns equal to the metadata columns plus the number of taxa. A bar chart shows sample counts before and after the join.

**How this helps the next stage:** All outlier exploration, diversity analysis, and ordination steps will use this single merged table, avoiding any synchronisation issues between separate DataFrames.

````python
print('=' * 60)
print('TASK 3: Merge Metadata and Taxonomic Profile')
print('=' * 60)

print(f'metadata_final shape:      {metadata_final.shape}')
print(f'tax_profile_samples shape: {tax_profile_samples.shape}')

merged_food_microbiome = pd.merge(
    metadata_final, tax_profile_samples,
    on='combined_id', how='inner'
)

n_meta  = metadata_final.shape[0]
n_tax   = tax_profile_samples.shape[0]
n_merge = merged_food_microbiome.shape[0]
n_lost  = n_meta - n_merge

print(f'\nmerged_food_microbiome shape: {merged_food_microbiome.shape}')
print(f'Matched samples:              {n_merge}')
print(f'Samples lost in merge:        {n_lost}')
if n_lost > 0:
    print('  (These samples existed in metadata but had no matching tax column, or vice versa.)')

display(merged_food_microbiome.iloc[:5, :8])

# Bar chart
fig, ax = plt.subplots(figsize=(8, 5))
bl2 = ['metadata_final\n(samples)', 'tax_profile_samples\n(samples)', 'merged_food_microbiome\n(samples)']
bv2 = [n_meta, n_tax, n_merge]
bc2 = ['#4C72B0', '#4C72B0', '#55A868']
bars2 = ax.bar(bl2, bv2, color=bc2, edgecolor='black', width=0.5)
for b, v in zip(bars2, bv2):
    ax.text(b.get_x() + b.get_width()/2, b.get_height() + max(bv2)*0.01,
             str(v), ha='center', va='bottom', fontweight='bold', fontsize=12)
ax.set_title('Sample Counts Before and After Merge', fontsize=13, fontweight='bold')
ax.set_ylabel('Number of Samples')
ax.set_ylim(0, max(bv2) * 1.15)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
````

### Interpretation — Merge Result

An inner join keeps only samples that appear in *both* tables. If the number of samples after the merge equals the number in `metadata_final` (and in `tax_profile_samples`), the two tables were perfectly aligned and no data was lost. A small number of dropped samples is acceptable and expected from the earlier duplicate-removal and matching steps. A large drop would indicate a key-format mismatch and would require investigation.

## Task 4 — Select Informative Taxa for Outlier Exploration

**What this step does:** Identifies the most informative bacterial taxa in `merged_food_microbiome` by computing mean abundance, variance, maximum value, and prevalence (the percentage of samples where the taxon is detected). The top 8 taxa by variance — among those present in at least 5% of samples — are selected for the scatter-plot analysis.

**Why it is needed:** The merged table contains thousands of taxa columns. Plotting all of them is impossible. Selecting taxa with high variance ensures we focus on the microorganisms that vary the most across samples, which are the most likely to show biologically meaningful patterns and outliers.

**What output to expect:** A summary table listing the selected taxa with their abundance statistics. This table makes the selection process transparent and reproducible.

**How this helps the next stage:** The selected taxa are used directly as the axes of the 10 scatter plots in Task 5.

````python
print('=' * 60)
print('TASK 4: Select Informative Taxa')
print('=' * 60)

META_COLS = ['combined_id', 'sample_id', 'fermented/non-fermented',
             'type', 'country', 'food_group']

tax_cols = [c for c in merged_food_microbiome.columns if c not in META_COLS]
print(f'Total taxonomic columns in merged table: {len(tax_cols)}')

# Numeric abundance matrix
tax_data = (
    merged_food_microbiome[tax_cols]
    .apply(pd.to_numeric, errors='coerce')
    .fillna(0)
)

# Per-taxon statistics
stats_df = pd.DataFrame({
    'taxon'         : tax_cols,
    'mean_abundance': tax_data.mean().round(6).values,
    'variance'      : tax_data.var().round(6).values,
    'max_value'     : tax_data.max().round(6).values,
    'prevalence_%'  : ((tax_data > 0).mean() * 100).round(2).values,
})

# Top 8: high variance AND present in at least 5 % of samples
informative = (
    stats_df[stats_df['prevalence_%'] > 5]
    .nlargest(8, 'variance')
    .reset_index(drop=True)
)

if len(informative) < 2:
    print('WARNING: Fewer than 2 taxa pass the prevalence threshold. '
          'Relaxing to top 8 by variance overall.')
    informative = stats_df.nlargest(8, 'variance').reset_index(drop=True)

def shorten(name, n=55):
    s = str(name)
    return s if len(s) <= n else '...' + s[-(n - 3):]

informative['short_name'] = informative['taxon'].apply(shorten)

print(f'\n{len(informative)} most informative taxa selected (ranked by variance, prevalence > 5%):')
display(
    informative[['short_name','mean_abundance','variance','max_value','prevalence_%']]
)
````

## Task 5 — Outlier Exploration: 10 Scatter Plots

**What this step does:** Creates up to 10 pairwise scatter plots using the informative taxa selected in Task 4. Each plot shows the co-abundance of two taxa across all food samples, coloured by `food_group`. IQR-based outliers are highlighted in red, and the most extreme outlier in each plot is annotated with its `combined_id`.

**Why it is needed:** Scatter plots are the most direct way to spot samples with unusual abundance patterns — either extremely high abundance in one taxon, or unexpected co-occurrence of high abundances in two taxa simultaneously. Both types of pattern can indicate contamination, mis-labelling, or genuinely interesting biological phenomena.

**What output to expect:** Exactly 10 scatter plots (or fewer if not enough taxa are available), each followed by a short printed interpretation. Red points mark IQR high outliers; the most extreme is labelled.

**Important:** This step does *not* run PCA, clustering, or statistical testing. It is purely visual, exploratory, and hypothesis-generating.

````python
selected_taxa = informative['taxon'].tolist()
n_sel = len(selected_taxa)

if n_sel < 2:
    print('WARNING: Not enough informative taxa to create scatter plots (need >= 2).')
else:
    pairs   = list(itertools.combinations(range(n_sel), 2))[:10]
    n_plots = len(pairs)
    if n_plots < 10:
        print(f'WARNING: Only {n_sel} taxa available — creating {n_plots} plots (< 10).')
    else:
        print(f'Creating {n_plots} scatter plots from {n_sel} selected taxa...')

    # Food-group colour palette
    fg_vals  = merged_food_microbiome['food_group'].dropna().unique()
    palette  = dict(zip(fg_vals, sns.color_palette('tab10', len(fg_vals))))

    def iqr_high_outliers(s):
        q1, q3 = s.quantile(0.25), s.quantile(0.75)
        return s > q3 + 1.5 * (q3 - q1)

    fg_series = merged_food_microbiome['food_group'].fillna('Unknown')

    for plot_num, (xi, yi) in enumerate(pairs, start=1):
        tx = selected_taxa[xi]
        ty = selected_taxa[yi]
        x  = tax_data[tx]
        y  = tax_data[ty]

        fig, ax = plt.subplots(figsize=(10, 6))

        # Scatter coloured by food_group
        for grp in fg_vals:
            m = fg_series == grp
            ax.scatter(x[m], y[m], c=[palette.get(grp, '#888888')],
                        alpha=0.5, edgecolors='none', s=45, label=grp)

        # Highlight IQR outliers
        out_mask = iqr_high_outliers(x) | iqr_high_outliers(y)
        n_out    = int(out_mask.sum())

        if 0 < n_out <= 20:
            ax.scatter(x[out_mask], y[out_mask],
                        c='red', alpha=0.85, edgecolors='black',
                        s=80, zorder=5, label=f'Outlier (n={n_out})')
            # Annotate the single most extreme point
            extreme_idx = (x + y).idxmax()
            ax.annotate(
                merged_food_microbiome.loc[extreme_idx, 'combined_id'],
                xy=(x[extreme_idx], y[extreme_idx]),
                xytext=(6, 4), textcoords='offset points',
                fontsize=7, color='darkred',
                arrowprops=dict(arrowstyle='->', color='darkred', lw=0.8)
            )

        ax.set_xlabel(f'Abundance: {shorten(tx, 50)}', fontsize=9)
        ax.set_ylabel(f'Abundance: {shorten(ty, 50)}', fontsize=9)
        ax.set_title(
            f'Scatter Plot {plot_num} of {n_plots} — Taxon Pair Outlier Check\n'
            f'Points coloured by food_group | Red = IQR high outliers',
            fontsize=12, fontweight='bold'
        )

        handles = [
            plt.Line2D([0],[0], marker='o', color='w',
                        markerfacecolor=palette.get(g,'#888888'),
                        markersize=8, label=g)
            for g in fg_vals
        ]
        if n_out > 0:
            handles.append(
                plt.Line2D([0],[0], marker='o', color='w',
                            markerfacecolor='red', markersize=8,
                            markeredgecolor='black', label=f'Outlier (n={n_out})')
            )
        ax.legend(handles=handles, bbox_to_anchor=(1.03, 1),
                   loc='upper left', fontsize=8, title='food_group')

        sns.despine(ax=ax)
        plt.tight_layout()
        plt.show()

        # Per-plot interpretation
        lx = shorten(tx, 40)
        ly = shorten(ty, 40)
        if n_out == 0:
            interp = ('No extreme IQR outliers detected for this pair. '
                       'Most samples share similar low abundance levels for both taxa, '
                       'suggesting they are neither universally dominant nor problematic.')
        elif n_out <= 5:
            interp = (f'{n_out} outlier(s) were detected (red). '
                       'These samples show unusually high abundance for at least one of '
                       'the two taxa and warrant biological review to determine whether '
                       'they represent a true biological signal or a data quality issue.')
        else:
            interp = (f'{n_out} samples fall outside the IQR boundaries for this pair. '
                       'The relatively high number of outliers suggests that one or both '
                       'taxa may be consistently elevated in a specific food group. '
                       'Inspect the colour pattern to see which food type drives the signal.')
        print(f'Plot {plot_num} interpretation: {interp}')
        print()
````

## Task 6 — Outlier Summary Table

**What this step does:** For each of the 8 selected taxa, computes Q1, Q3, IQR, and the high-outlier threshold (Q3 + 1.5 × IQR). Samples whose abundance exceeds this threshold are flagged. The results are shown in two tables: a per-taxon summary and a detailed list of the top outlier samples.

**Why it is needed:** Scatter plots give a visual impression of outliers, but a quantitative table makes it possible to act on the findings: decide which samples to investigate further, flag for biological review, or potentially remove before modelling.

**What output to expect:** A summary table (one row per taxon) and a detailed table showing the top 20 outlier sample records including their metadata.

**Important note:** These are *exploratory* outliers identified by a statistical rule. They are not necessarily errors. High-abundance readings may reflect genuine biological differences — for example, a cheese sample dominated by a single starter culture. Biological context must be considered before any sample is removed.

````python
print('=' * 60)
print('TASK 6: IQR-Based Outlier Summary')
print('=' * 60)

summary_rows = []
outlier_dfs  = []

for _, row in informative.iterrows():
    taxon = row['taxon']
    short = row['short_name']
    vals  = tax_data[taxon]

    q1  = vals.quantile(0.25)
    q3  = vals.quantile(0.75)
    iqr = q3 - q1
    hi  = q3 + 1.5 * iqr

    out_mask = vals > hi
    n_out    = int(out_mask.sum())

    summary_rows.append({
        'Taxon (truncated)'       : short,
        'Q1'                      : round(float(q1), 5),
        'Q3'                      : round(float(q3), 5),
        'IQR'                     : round(float(iqr), 5),
        'High Threshold (Q3+1.5xIQR)': round(float(hi), 5),
        'N High Outliers'         : n_out,
    })

    if n_out > 0:
        out_df = merged_food_microbiome.loc[out_mask,
            ['combined_id','sample_id','food_group','type','country']
        ].copy()
        out_df['taxon']     = short
        out_df['abundance'] = vals[out_mask].values
        outlier_dfs.append(out_df)

summary_tbl = pd.DataFrame(summary_rows)
print('\nOutlier Count per Selected Taxon:')
display(summary_tbl)

if outlier_dfs:
    detailed = (
        pd.concat(outlier_dfs, ignore_index=True)
        .sort_values('abundance', ascending=False)
        .head(20)
        .reset_index(drop=True)
    )
    total_events = sum(r['N High Outliers'] for r in summary_rows)
    print(f'\nTop 20 Outlier Sample Records'
          f' (from {total_events} total outlier events across {len(informative)} taxa):')
    display(detailed)
else:
    print('\nNo high outliers detected for any of the selected taxa.')
````

# Taxonomic Rank Aggregation on the Merged Food Microbiome Table

This section takes the existing `merged_food_microbiome` table and adds new aggregated abundance columns for every taxonomic rank — kingdom, phylum, class, order, family, genus, species, and strain.

Each new column holds the **row-wise sum** of all original taxon columns that share the same value at a given rank.  For example, the single column `rank_p__Firmicutes` will equal the sum of every taxon column whose path contains `p__Firmicutes`.

**Input:** `merged_food_microbiome`  
**Key output:** `merged_food_microbiome_with_rank_sums`  
**No files are written to disk in this section.**

## Task 1 — Identify Taxonomic Columns

**What this step does:** Separates the columns of `merged_food_microbiome` into two groups — metadata columns and taxonomic abundance columns — by checking whether each column name contains a recognised rank prefix (`k__`, `p__`, `c__`, `o__`, `f__`, `g__`, `s__`, or `t__`).

**Why it is needed:** Subsequent parsing and aggregation steps should only operate on genuine taxonomic path strings.  Accidentally including metadata columns (such as `food_group` or `country`) in the aggregation would produce incorrect sums.

**What output to expect:** Printed counts of metadata and taxonomic columns, plus a few example taxonomic column names to confirm the detection is correct.

**How this helps the next stage:** The list `tax_cols` is the input to the parsing step and to all aggregation operations that follow.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

print('=' * 60)
print('TASK 1: Identify Taxonomic Columns')
print('=' * 60)

# Metadata columns that must never be treated as taxa
META_COLS = [
    'combined_id', 'sample_id', 'fermented/non-fermented',
    'type', 'country', 'food_group'
]

RANK_PREFIXES = ['k__', 'p__', 'c__', 'o__', 'f__', 'g__', 's__', 't__']

def is_taxon_col(col):
    """Returns True if the column name looks like a taxonomic path."""
    return any(pfx in col for pfx in RANK_PREFIXES)

all_cols   = merged_food_microbiome.columns.tolist()
tax_cols   = [c for c in all_cols if c not in META_COLS and is_taxon_col(c)]
meta_cols  = [c for c in all_cols if c not in tax_cols]

print(f'Total columns in merged_food_microbiome : {len(all_cols)}')
print(f'  Metadata columns detected             : {len(meta_cols)}')
print(f'  Taxonomic columns detected            : {len(tax_cols)}')

print('\nExample taxonomic column names (first 5):')
for ex in tax_cols[:5]:
    print(f'  {ex}')

print('\nExample taxonomic column names (last 5):')
for ex in tax_cols[-5:]:
    print(f'  {ex}')
````

## Task 2 — Parse Taxonomic Paths into a Mapping Table

**What this step does:** For every taxonomic column, extracts the value at each rank level (k, p, c, o, f, g, s, t) by scanning the column name for the corresponding prefix pattern.  The results are stored in a DataFrame called `taxonomic_column_mapping`.

**Why it is needed:** Without this mapping, we would have to re-parse each column name repeatedly during aggregation.  Doing the parsing once and storing it in a lookup table is both faster and easier to audit.

**What output to expect:** A preview of `taxonomic_column_mapping` (one row per taxon column, one column per rank) and a count of unique values found at each rank.

**How this helps the next stage:** `taxonomic_column_mapping` is the backbone of the aggregation step — it tells us exactly which original columns belong to each rank value.

````python
print('=' * 60)
print('TASK 2: Parse Taxonomic Paths')
print('=' * 60)

# Map rank short names to display names and column-prefix strings
RANK_INFO = [
    ('kingdom', 'k__'),
    ('phylum',  'p__'),
    ('class',   'c__'),
    ('order',   'o__'),
    ('family',  'f__'),
    ('genus',   'g__'),
    ('species', 's__'),
    ('strain',  't__'),
]

def parse_taxon_path(col_name):
    result = {}
    for rank_name, prefix in RANK_INFO:
        # Extract the value immediately after the prefix, up to the next '|'
        m = re.search(re.escape(prefix) + r'([^|]+)', col_name)
        result[rank_name] = m.group(1) if m else 'Unclassified'
    return result

# Build the mapping table (one row per taxon column)
rows = []
for col in tax_cols:
    parsed = parse_taxon_path(col)
    parsed['original_taxon_column'] = col
    rows.append(parsed)

taxonomic_column_mapping = pd.DataFrame(rows, columns=
    ['original_taxon_column'] + [r for r, _ in RANK_INFO]
)

print(f'taxonomic_column_mapping shape: {taxonomic_column_mapping.shape}')
print('\nPreview:')
display(taxonomic_column_mapping.head())

print('\nUnique values per rank:')
for rank_name, _ in RANK_INFO:
    n_unique = taxonomic_column_mapping[rank_name].nunique()
    print(f'  {rank_name:10s}: {n_unique:5d} unique values')
````

## Task 3 — Aggregate Abundance by Rank

**What this step does:** Creates `merged_food_microbiome_with_rank_sums` as a copy of `merged_food_microbiome` and then, for each rank and each unique value within that rank, appends one new column whose value is the row-wise sum of all original taxon columns that share that rank value.

**Why it is needed:** The raw table has thousands of fine-grained taxon columns.  Rank-level aggregates let us analyse microbial diversity at biologically meaningful resolutions — for example, phylum-level proportions are widely used to characterise microbiome composition, and genus-level profiles are the standard input for food-safety and quality studies.

**What output to expect:** For each rank the code prints the number of new columns created and (if the rank was skipped because it would add too many columns) a clear warning.  A rank with many unique values (e.g. strain) may be skipped automatically to avoid creating an impractically wide table.

**How this helps the next stage:** The new `rank_*` columns can be used directly for grouped bar plots, heatmaps, ANOVA, and ordination without any further parsing.

````python
print('=' * 60)
print('TASK 3: Aggregate Abundance by Rank')
print('=' * 60)

# If a rank has more than this many unique values, aggregating it would
# add an impractically large number of new columns (essentially duplicating
# the original taxon columns at strain level). Those ranks are skipped.
MAX_UNIQUE = 1500

merged_food_microbiome_with_rank_sums = merged_food_microbiome.copy()

# Pre-convert the taxon block to numeric once for speed
tax_matrix = (
    merged_food_microbiome[tax_cols]
    .apply(pd.to_numeric, errors='coerce')
    .fillna(0)
)

new_cols_per_rank = {}   # rank_name -> list of new column names added
skipped_ranks     = set()

for rank_name, prefix in RANK_INFO:
    n_unique = taxonomic_column_mapping[rank_name].nunique()

    if n_unique > MAX_UNIQUE:
        print(f'  [{rank_name}] SKIPPED — {n_unique} unique values '
              f'(threshold = {MAX_UNIQUE}). Adding {n_unique} columns '
              f'at strain level is equivalent to duplicating the original '
              f'taxon columns and adds little analytical value.')
        new_cols_per_rank[rank_name] = []
        skipped_ranks.add(rank_name)
        continue

    added = []
    groups = taxonomic_column_mapping.groupby(rank_name)['original_taxon_column'].apply(list)

    for val, cols_in_group in groups.items():
        col_name     = f'rank_{prefix}{val}'
        valid_cols   = [c for c in cols_in_group if c in tax_matrix.columns]
        if valid_cols:
            merged_food_microbiome_with_rank_sums[col_name] = (
                tax_matrix[valid_cols].sum(axis=1)
            )
            added.append(col_name)

    new_cols_per_rank[rank_name] = added
    print(f'  [{rank_name}] {n_unique:4d} unique values  ->  {len(added):4d} new columns added')

total_new = sum(len(v) for v in new_cols_per_rank.values())
print(f'\nTotal new rank-sum columns added: {total_new}')
print(f'merged_food_microbiome_with_rank_sums shape: '
      f'{merged_food_microbiome_with_rank_sums.shape}')
````

## Task 4 — Validate Rank-Sum Columns

**What this step does:** For each rank, sums all newly created `rank_*` columns row-wise and compares the result to the total taxonomic abundance of each sample (the row-wise sum of the original taxon columns).  If the aggregation is correct, the two sums should match for every sample.

**Why it is needed:** Summing columns is error-prone: a typo in a column name or an accidental duplication would silently inflate the rank totals.  This validation step catches such mistakes before they propagate.

**What output to expect:** A table with one row per rank showing the minimum, mean, and maximum of the per-sample rank-sum, plus a count of samples that deviate from the expected total by more than 1%.  All counts should ideally be 0.

**How this helps the next stage:** Confirms that the new columns can be used for quantitative comparisons without bias introduced by aggregation errors.

````python
print('=' * 60)
print('TASK 4: Validate Rank-Sum Columns')
print('=' * 60)

# Row-wise total of all original taxon columns
total_abundance = tax_matrix.sum(axis=1)
mean_total = float(total_abundance.mean())

# Detect scale: proportions (~1) vs percentages (~100)
if mean_total > 10:
    scale_label = 'percentage scale (~100 per sample)'
else:
    scale_label = 'proportion scale (~1 per sample)'
print(f'Detected abundance scale: {scale_label}')
print(f'Mean total abundance per sample: {mean_total:.4f}')

tolerance_pct = 0.01   # 1 % relative tolerance

val_rows = []
for rank_name, _ in RANK_INFO:
    new_cols = new_cols_per_rank.get(rank_name, [])
    if not new_cols:
        val_rows.append({
            'Rank': rank_name,
            'New Cols': 0,
            'Min row-sum': 'SKIPPED',
            'Mean row-sum': 'SKIPPED',
            'Max row-sum': 'SKIPPED',
            'Samples off (>1%)': 'SKIPPED',
        })
        continue

    rank_total = merged_food_microbiome_with_rank_sums[new_cols].sum(axis=1)
    diff       = (rank_total - total_abundance).abs()
    tol        = total_abundance * tolerance_pct
    n_off      = int((diff > tol).sum())

    val_rows.append({
        'Rank': rank_name,
        'New Cols': len(new_cols),
        'Min row-sum': round(float(rank_total.min()), 5),
        'Mean row-sum': round(float(rank_total.mean()), 5),
        'Max row-sum': round(float(rank_total.max()), 5),
        'Samples off (>1%)': n_off,
    })

val_df = pd.DataFrame(val_rows)
display(val_df)

n_failed = sum(
    int(r['Samples off (>1%)']) for r in val_rows
    if r['Samples off (>1%)'] != 'SKIPPED'
)
if n_failed == 0:
    print('\nValidation PASSED: all rank sums match the original total within 1%.')
else:
    print(f'\nValidation WARNING: {n_failed} sample(s) deviate by more than 1%.')
    print('Check whether any taxon columns contain NaN or non-numeric values.')
````

### Note on Floating-Point Differences

Very small discrepancies between rank sums and the original total (typically less than 1e-10) are normal and caused by the limited precision of 64-bit floating-point arithmetic.  When many numbers near zero are added together, rounding errors accumulate.  These are **not** data quality issues.  The 1% tolerance used above is far wider than any floating-point error and will only flag genuine mistakes in the aggregation logic.

**What this step does:** Applies a regular-expression parser to every taxonomic column name to extract the value at each rank level (kingdom, phylum, class, order, family, genus, species, strain). If a rank is absent from a column's path, it is recorded as `Unclassified`. The result is stored in a DataFrame called `taxonomic_column_mapping`.

**Why it is needed:** Parsing each column name once and caching the result in a lookup table is far more efficient than re-parsing during the aggregation loop. It also makes the mapping fully auditable.

**What output to expect:** A head preview of `taxonomic_column_mapping` (one row per original taxon column, one column per rank) and a summary of how many unique groups exist at each rank.

**How this helps the next stage:** `taxonomic_column_mapping` is the backbone of Task 3 — it tells the aggregation loop exactly which original columns belong to each rank value.

````python
print('=' * 60)
print('TASK 2: Parse Taxonomic Paths')
print('=' * 60)

RANK_INFO = [
    ('kingdom', 'k__'),
    ('phylum',  'p__'),
    ('class',   'c__'),
    ('order',   'o__'),
    ('family',  'f__'),
    ('genus',   'g__'),
    ('species', 's__'),
    ('strain',  't__'),
]

def parse_taxon_path(col_name):
    result = {}
    for rank_name, prefix in RANK_INFO:
        m = re.search(re.escape(prefix) + r'([^|]+)', col_name)
        result[rank_name] = m.group(1) if m else 'Unclassified'
    return result

rows = []
for col in tax_cols:
    parsed = parse_taxon_path(col)
    parsed['original_taxon_column'] = col
    rows.append(parsed)

col_order = ['original_taxon_column'] + [r for r, _ in RANK_INFO]
taxonomic_column_mapping = pd.DataFrame(rows, columns=col_order)

print(f'taxonomic_column_mapping shape: {taxonomic_column_mapping.shape}')
print('\nPreview (first 5 rows):')
display(taxonomic_column_mapping.head())

print('\nUnique values per rank:')
for rank_name, _ in RANK_INFO:
    n = taxonomic_column_mapping[rank_name].nunique()
    print(f'  {rank_name:10s}: {n:6d} unique values')
````

## Task 3 — Aggregate Abundance by Rank → `merged_food_microbiome_with_rank_sums`

**What this step does:** Creates `merged_food_microbiome_with_rank_sums` as a copy of `merged_food_microbiome`, then for every rank and every unique value at that rank, appends one new column whose value is the row-wise sum of all original taxon columns sharing that rank value.

For example: if 200 original columns contain `p__Firmicutes` in their path, a single new column `rank_p__Firmicutes` is created whose value per sample equals the sum of those 200 columns.

**Why it is needed:** Working with 5,000 individual taxon columns is impractical for plotting and modelling. Rank-level aggregates provide a biologically meaningful and computationally manageable representation.

**What output to expect:** Progress lines showing how many columns were added at each rank. Ranks with more unique values than a configurable threshold (`MAX_UNIQUE`) are skipped to avoid creating an impractically wide table — a clear warning is printed for any skipped rank.

**How this helps the next stage:** The new `rank_*` columns are ready for grouped bar charts, heatmaps, and feature selection without further parsing.

````python
print('=' * 60)
print('TASK 3: Aggregate Abundance by Rank')
print('=' * 60)

# Ranks with more unique values than this are skipped.
# Strain-level aggregation typically produces as many new columns as there
# are original taxon columns, effectively doubling the table width.
# Lower this value to save memory; raise it to include more ranks.
MAX_UNIQUE = 2000

# Start from a clean copy — original table is never modified
merged_food_microbiome_with_rank_sums = merged_food_microbiome.copy()

# Convert the entire taxon block to numeric once for speed
tax_matrix = (
    merged_food_microbiome[tax_cols]
    .apply(pd.to_numeric, errors='coerce')
    .fillna(0)
)

new_cols_per_rank = {}   # rank_name  -> list of new column names
skipped_ranks     = set()

for rank_name, prefix in RANK_INFO:
    n_unique = taxonomic_column_mapping[rank_name].nunique()

    if n_unique > MAX_UNIQUE:
        print(f'  [{rank_name:10s}] SKIPPED — {n_unique} unique values '
              f'exceeds MAX_UNIQUE={MAX_UNIQUE}. '
              f'Raise MAX_UNIQUE if you need strain-level columns.')
        new_cols_per_rank[rank_name] = []
        skipped_ranks.add(rank_name)
        continue

    groups = (
        taxonomic_column_mapping
        .groupby(rank_name)['original_taxon_column']
        .apply(list)
    )
    added = []
    for val, cols_in_group in groups.items():
        new_col = f'rank_{prefix}{val}'
        valid   = [c for c in cols_in_group if c in tax_matrix.columns]
        if valid:
            merged_food_microbiome_with_rank_sums[new_col] = (
                tax_matrix[valid].sum(axis=1)
            )
            added.append(new_col)

    new_cols_per_rank[rank_name] = added
    print(f'  [{rank_name:10s}] {n_unique:5d} unique values  '
          f'->  {len(added):5d} new columns added')

total_new = sum(len(v) for v in new_cols_per_rank.values())
print(f'\nTotal new rank-sum columns added        : {total_new}')
print(f'Original merged_food_microbiome columns : {len(merged_food_microbiome.columns)}')
print(f'New table total columns                 : '
      f'{len(merged_food_microbiome_with_rank_sums.columns)}')
print(f'Shape: {merged_food_microbiome_with_rank_sums.shape}')
````

## Task 4 — Validate Rank-Sum Columns

**What this step does:** For each processed rank, sums all newly created `rank_*` columns row-wise per sample and compares the result to the row-wise total of the original taxon columns. Since every original taxon column belongs to exactly one value at each rank, the two sums should match for every sample.

**Why it is needed:** A summing error — caused by a mis-parsed column name, a typo, or an accidental duplication — would silently inflate or deflate the rank totals. This step catches such mistakes before they propagate.

**What output to expect:** A validation table with one row per rank showing the min, mean, and max row-sum and the count of samples that deviate from the original total by more than 1%. All deviation counts should be 0.

**Note on floating-point differences:** Aggregating thousands of very small floating-point numbers introduces rounding errors at the level of 1e-12 to 1e-10. These are completely normal and are far smaller than the 1% tolerance used here. They are **not** data quality issues.

````python
print('=' * 60)
print('TASK 4: Validate Rank-Sum Columns')
print('=' * 60)

total_abundance = tax_matrix.sum(axis=1)
mean_total      = float(total_abundance.mean())
scale_label     = 'percentage scale (~100)' if mean_total > 10 else 'proportion scale (~1)'
print(f'Detected abundance scale: {scale_label}  (mean per-sample total = {mean_total:.5f})')

TOLERANCE = 0.01   # 1 % relative tolerance

val_rows = []
for rank_name, _ in RANK_INFO:
    cols = new_cols_per_rank.get(rank_name, [])
    if not cols:
        val_rows.append(dict(Rank=rank_name, New_Cols=0,
                              Min='SKIPPED', Mean='SKIPPED',
                              Max='SKIPPED', Samples_off='SKIPPED'))
        continue

    rank_total = merged_food_microbiome_with_rank_sums[cols].sum(axis=1)
    diff       = (rank_total - total_abundance).abs()
    tol        = total_abundance.abs() * TOLERANCE
    n_off      = int((diff > tol).sum())

    val_rows.append(dict(
        Rank        = rank_name,
        New_Cols    = len(cols),
        Min         = round(float(rank_total.min()),  6),
        Mean        = round(float(rank_total.mean()), 6),
        Max         = round(float(rank_total.max()),  6),
        Samples_off = n_off,
    ))

val_df = pd.DataFrame(val_rows).rename(columns={
    'New_Cols': 'New Cols Added',
    'Min': 'Min row-sum', 'Mean': 'Mean row-sum', 'Max': 'Max row-sum',
    'Samples_off': 'Samples off (>1%)'
})
print('\nValidation summary:')
display(val_df)

n_failed = sum(
    r['Samples_off'] for r in val_rows
    if isinstance(r['Samples_off'], int)
)
if n_failed == 0:
    print('\nValidation PASSED: all rank sums match the original total within 1%.')
else:
    print(f'\nValidation WARNING: {n_failed} sample(s) deviate by more than 1%.')
    print('Check for NaN or non-numeric values in the original taxon columns.')
````

## Task 5 — Visualisations

**What this step does:** Produces eight charts that together explain the structural outcome of the aggregation and the biological content of the dominant microbial groups.

**Why it is needed:** Numbers in a validation table cannot communicate the relative importance of phyla, families, or genera. Visual summaries make it immediately clear which microbial groups dominate and how composition differs across food types.

**What output to expect:**
- Two structural overview bar charts (unique values per rank, new columns per rank)
- Four top-15 abundance charts (phyla, classes, families, genera)
- Two food-group heatmaps (top phyla × food_group, top genera × food_group)

**How this helps the next stage:** These charts are presentation-ready and provide biological context for any diversity or modelling analysis that follows.

````python
# ── Fig A: unique categories per rank + new columns per rank ─────────────
ranks_done    = [r for r, _ in RANK_INFO if new_cols_per_rank.get(r)]
n_unique_list = [taxonomic_column_mapping[r].nunique() for r in ranks_done]
n_new_list    = [len(new_cols_per_rank[r])              for r in ranks_done]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

bars_a = axes[0].bar(ranks_done, n_unique_list, color='#4C72B0', edgecolor='black')
for b, v in zip(bars_a, n_unique_list):
    axes[0].text(b.get_x() + b.get_width()/2,
                  b.get_height() + max(n_unique_list) * 0.01,
                  str(v), ha='center', va='bottom', fontweight='bold', fontsize=10)
axes[0].set_title('Unique Taxonomic Values per Rank\n(processed ranks only)',
                   fontsize=12, fontweight='bold')
axes[0].set_xlabel('Rank'); axes[0].set_ylabel('Unique Values')
axes[0].tick_params(axis='x', rotation=15)
sns.despine(ax=axes[0])

bars_b = axes[1].bar(ranks_done, n_new_list, color='#55A868', edgecolor='black')
for b, v in zip(bars_b, n_new_list):
    axes[1].text(b.get_x() + b.get_width()/2,
                  b.get_height() + max(n_new_list) * 0.01,
                  str(v), ha='center', va='bottom', fontweight='bold', fontsize=10)
axes[1].set_title('New rank_* Columns Added per Rank',
                   fontsize=12, fontweight='bold')
axes[1].set_xlabel('Rank'); axes[1].set_ylabel('New Columns')
axes[1].tick_params(axis='x', rotation=15)
sns.despine(ax=axes[1])

plt.suptitle('Aggregation Overview: Structure of What Was Added',
              fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print('Interpretation: Higher ranks (kingdom, phylum) have fewer unique values '
      'and therefore add fewer, broader columns. Lower ranks (genus, species) '
      'have many more unique values, producing a finer-grained but wider set of '
      'aggregate columns. The two charts should always show the same bar heights '
      '— confirming that one new column was created per unique rank value.')
````

````python
# ── Fig B: top 15 groups by mean abundance — phylum, class, family, genus ─

def top15_bar(ax, rank_name, prefix, color, title):
    cols = new_cols_per_rank.get(rank_name, [])
    if not cols:
        ax.text(0.5, 0.5, f'{rank_name} was skipped.',
                 ha='center', va='center', fontsize=11, transform=ax.transAxes)
        ax.set_title(title, fontsize=11, fontweight='bold')
        ax.axis('off')
        return
    means = (
        merged_food_microbiome_with_rank_sums[cols]
        .mean().sort_values(ascending=False).head(15)
    )
    labels = [c.replace(f'rank_{prefix}', '') for c in means.index]
    sns.barplot(x=means.values, y=labels, ax=ax, color=color)
    mx = float(means.values.max())
    for i, v in enumerate(means.values):
        ax.text(v + mx * 0.01, i,
                 f'{v:.5f}', va='center', fontsize=8)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.set_xlabel('Mean Abundance'); ax.set_ylabel('')
    sns.despine(ax=ax)

fig, axes = plt.subplots(2, 2, figsize=(18, 14))
top15_bar(axes[0,0], 'phylum', 'p__', '#4C72B0', 'Top 15 Phyla by Mean Abundance')
top15_bar(axes[0,1], 'class',  'c__', '#DD8452', 'Top 15 Classes by Mean Abundance')
top15_bar(axes[1,0], 'family', 'f__', '#55A868', 'Top 15 Families by Mean Abundance')
top15_bar(axes[1,1], 'genus',  'g__', '#C44E52', 'Top 15 Genera by Mean Abundance')
plt.suptitle('Top 15 Taxonomic Groups by Mean Abundance Across All Samples',
              fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print('Interpretation: A long bar indicates a taxon that is both highly abundant '
      'and consistently present across many food samples. Dominant phyla '
      '(e.g. Firmicutes, Proteobacteria) set the broad microbial context, while '
      'the genus chart reveals the specific bacteria driving that signal — '
      'typically Lactobacillus, Lactococcus, and related lactic acid bacteria '
      'for fermented food datasets.')
````

````python
# ── Fig C: heatmaps — top phyla and genera across food groups ─────────────

if 'food_group' not in merged_food_microbiome_with_rank_sums.columns:
    print('food_group column not found — food-group heatmaps skipped.')
else:
    def rank_heatmap(rank_name, prefix, n_top, title):
        cols = new_cols_per_rank.get(rank_name, [])
        if not cols:
            print(f'{rank_name} was skipped — heatmap not available.')
            return
        top_cols = (
            merged_food_microbiome_with_rank_sums[cols]
            .mean().nlargest(n_top).index.tolist()
        )
        heat_df = (
            merged_food_microbiome_with_rank_sums
            .dropna(subset=['food_group'])
            .groupby('food_group')[top_cols]
            .mean()
        )
        heat_df.columns = [c.replace(f'rank_{prefix}', '') for c in heat_df.columns]
        max_val = float(heat_df.values.max())
        fmt     = '.4f' if max_val < 0.1 else ('.2f' if max_val < 1 else '.1f')
        fig_h, ax_h = plt.subplots(
            figsize=(max(10, n_top * 1.0), max(5, len(heat_df) * 0.6))
        )
        sns.heatmap(
            heat_df, ax=ax_h, cmap='YlOrRd',
            annot=True, fmt=fmt, linewidths=0.4,
            cbar_kws={'label': 'Mean Abundance'}
        )
        ax_h.set_title(title, fontsize=13, fontweight='bold', pad=14)
        ax_h.set_xlabel(''); ax_h.set_ylabel('Food Group')
        plt.xticks(rotation=40, ha='right', fontsize=9)
        plt.yticks(rotation=0, fontsize=9)
        plt.tight_layout()
        plt.show()

    rank_heatmap('phylum', 'p__', 10,
                  'Top 10 Phyla — Mean Abundance per Food Group')
    print('Interpretation (Phyla heatmap): Warm cells (yellow-red) indicate '
          'higher mean phylum abundance for that food group. Fermented dairy '
          'is typically dominated by Firmicutes; environmental or plant-based '
          'foods tend to show more even phylum profiles. Cold cells near zero '
          'mean that phylum is essentially absent in that food type.\n')

    rank_heatmap('genus', 'g__', 12,
                  'Top 12 Genera — Mean Abundance per Food Group')
    print('Interpretation (Genera heatmap): Genus-level signals are the most '
          'informative for food-type classification. Food-group-specific hot '
          'spots (e.g. Lactococcus in cheese, Acetobacter in vinegar) confirm '
          'that food_group captures real biological variation and will be a '
          'strong predictive variable in downstream modelling.')
````

## Task 6 — Final Output and Rank-Level Statistics Visualisation

**What this step does:** Prints a complete shape and column-count summary, previews the new table with both metadata and rank-sum columns visible side by side, and produces a statistics visualisation showing the mean, standard deviation, and spread of abundance values across all new rank-sum columns at each rank level.

**Why it is needed:** Confirming the final table dimensions and previewing representative columns closes the loop on the aggregation work and provides a quantitative picture of how microbial abundance is distributed across taxonomic resolutions.

**What output to expect:** A summary table, a head preview, a bar chart with error bars showing mean ± SD per rank, and a printed narrative summary.

**How this helps the next stage:** The visualisation immediately shows whether higher ranks (phyla) or lower ranks (genera) drive more variance across samples — informing which rank level to prioritise in downstream EDA.

````python
print('=' * 60)
print('TASK 6: Final Output')
print('=' * 60)

n_orig    = len(merged_food_microbiome.columns)
n_new_tot = sum(len(v) for v in new_cols_per_rank.values())
n_final   = len(merged_food_microbiome_with_rank_sums.columns)

# ── Shape summary table ───────────────────────────────────────────────────
summary_rows = [
    ['merged_food_microbiome — rows',                    merged_food_microbiome.shape[0]],
    ['merged_food_microbiome — columns',                 n_orig],
    ['New rank-sum columns added',                       n_new_tot],
    ['merged_food_microbiome_with_rank_sums — rows',     merged_food_microbiome_with_rank_sums.shape[0]],
    ['merged_food_microbiome_with_rank_sums — columns',  n_final],
]
for r, _ in RANK_INFO:
    nc     = len(new_cols_per_rank.get(r, []))
    status = 'SKIPPED' if r in skipped_ranks else f'{nc} new columns'
    summary_rows.append([f'  rank level: {r}', status])

summary_df = pd.DataFrame(summary_rows, columns=['Metric', 'Value'])
print('\nShape & Column Summary:')
display(summary_df)

# ── Head preview ─────────────────────────────────────────────────────────
preview_cols = (
    list(merged_food_microbiome.columns[:4]) +
    new_cols_per_rank.get('phylum',  [])[:3] +
    new_cols_per_rank.get('genus',   [])[:3]
)
preview_cols = [c for c in preview_cols
                if c in merged_food_microbiome_with_rank_sums.columns]
print('\nHead preview (metadata + sample phylum & genus rank-sum columns):')
display(merged_food_microbiome_with_rank_sums[preview_cols].head())

# ── Stats averages per rank ───────────────────────────────────────────────
stat_rows = []
for rank_name, _ in RANK_INFO:
    cols = new_cols_per_rank.get(rank_name, [])
    if not cols:
        continue
    vals = merged_food_microbiome_with_rank_sums[cols].values.flatten()
    stat_rows.append({
        'rank'  : rank_name,
        'mean'  : round(float(np.mean(vals)),   8),
        'std'   : round(float(np.std(vals)),    8),
        'median': round(float(np.median(vals)), 8),
        'max'   : round(float(np.max(vals)),    8),
    })

stats_df = pd.DataFrame(stat_rows)
print('\nRank-level abundance statistics (across all new rank-sum columns):')
display(stats_df)

# ── Bar chart: mean ± SD per rank ─────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 5))
x_pos  = range(len(stats_df))
colors = plt.cm.tab10.colors[:len(stats_df)]
bars_s = ax.bar(
    stats_df['rank'], stats_df['mean'],
    color=colors, edgecolor='black', width=0.6
)
ax.errorbar(
    stats_df['rank'], stats_df['mean'],
    yerr=stats_df['std'],
    fmt='none', color='black', capsize=6, linewidth=1.8, elinewidth=1.5
)
for b, (_, row) in zip(bars_s, stats_df.iterrows()):
    ax.text(
        b.get_x() + b.get_width()/2,
        b.get_height() + float(stats_df['std'].max()) * 0.05,
        f'{row["mean"]:.5f}',
        ha='center', va='bottom', fontsize=8, fontweight='bold'
    )
ax.set_title(
    'Mean Abundance per Rank-Sum Column\n'
    '(error bars = ±1 standard deviation across all rank-sum columns at that rank)',
    fontsize=12, fontweight='bold'
)
ax.set_xlabel('Taxonomic Rank')
ax.set_ylabel('Mean Abundance Value')
ax.tick_params(axis='x', rotation=15)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()

print('Interpretation: Higher-rank columns (phylum, class) show larger mean '
      'values because each column aggregates many fine-grained taxon columns. '
      'The standard deviation (error bar) tells us how variable the total '
      'abundance is across the groups within that rank — a large SD at genus '
      'level means some genera are far more abundant than others, which is '
      'exactly what we expect in a food-microbiome dataset where a few '
      'dominant genera (Lactobacillus, Lactococcus) dominate.')

# ── Narrative summary ─────────────────────────────────────────────────────
val_ok = not any(
    isinstance(r.get('Samples_off'), int) and r['Samples_off'] > 0
    for r in val_rows
)
print(f'\n{"-" * 55}')
print('SECTION SUMMARY')
print(f'{"-" * 55}')
print(f'  Taxonomic columns parsed            : {len(tax_cols)}')
print(f'  New rank-sum columns added          : {n_new_tot}')
print(f'  Ranks skipped (> MAX_UNIQUE values) : {sorted(skipped_ranks) or "none"}')
print(f'  Rank-sum validation passed          : {val_ok}')
print(f'  Final table shape                   : {merged_food_microbiome_with_rank_sums.shape}')
print()
print('Why merged_food_microbiome_with_rank_sums is useful for later EDA:')
print('  rank_p__* — phylum-level diversity comparisons across food groups')
print('  rank_g__* — genus-level profiles, the standard input for food classification')
print('  rank_f__* — family profiles bridge species detail with phylum context')
print('  rank_o__* — order-level features useful for broad ecological comparisons')
print('  All rank columns are pre-computed: no repeated regex parsing in later cells')
````
